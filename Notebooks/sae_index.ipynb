```python
import torch

# === CONFIG — edit here ===
VECTOR_PATH = "Vector/SRPS/Gemma/deception"  # best SPARE result
LAY = 14  # layer
TOP_N = 15  # how many top features to show
# ===========================

metadata = torch.load(f"{VECTOR_PATH}/metadata.pt", map_location="cpu", weights_only=False)
top_idx_pos = metadata.get("top_idx", {}).get(LAY, [])
top_idx_neg = metadata.get("top_idx_neg", {}).get(LAY, [])

print(f"Method : {metadata.get('method', 'unknown')}")
print(f"Layer  : {LAY}")
print(f"Top pos features ({len(top_idx_pos)}): {top_idx_pos[:TOP_N]}")
print(f"Top neg features ({len(top_idx_neg)}): {top_idx_neg[:TOP_N]}")
```

```python
from IPython.display import IFrame, display

html_template = "https://neuronpedia.org/{}/{}/{}?embed=true&embedexplanation=true&embedplots=true&embedtest=true&height=300"

def show_neuronpedia(feature_idx: int, layer: int, model="gemma"):
    """Display a feature on Neuronpedia."""
    if model == "gemma":
        html = html_template.format("gemma-2-2b", f"{layer}-gemmascope-res-16k", feature_idx)
    else:
        html = html_template.format("llama3.1-8b", f"{layer}-llamascope-res-32k", feature_idx)
    return IFrame(html, width=1200, height=300)

# --- Positive features (contextual / target behaviour) ---
print(f"=== TOP {TOP_N} POSITIVE features (layer {LAY}) ===")
for idx in top_idx_pos[:TOP_N]:
    print(f"Feature {idx}")
    display(show_neuronpedia(int(idx), LAY, model="gemma"))

# --- Negative features (parametric / contrast behaviour) ---
print(f"=== TOP {TOP_N} NEGATIVE features (layer {LAY}) ===")
for idx in top_idx_neg[:TOP_N]:
    print(f"Feature {idx}")
    display(show_neuronpedia(int(idx), LAY, model="gemma"))
```
